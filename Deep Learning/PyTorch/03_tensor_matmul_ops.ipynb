환경: `torch-env` 커널을 쓰는 것을 권장합니다.

## 1. 준비 — import

```python
import torch
import numpy as np

print("torch 버전:", torch.__version__)
```

## 2. NumPy — 벡터 @ 행렬

길이 3인 벡터와 (3, 2) 행렬을 곱하면 결과는 길이 2인 벡터입니다.

```python
a = np.array([1, 2, 3])
B = np.array([[4, 5], [6, 7], [8, 9]])
result = a @ B
print(result)
print(result.shape)
```

## 3. PyTorch matmul — 행렬 × 벡터

(3, 2) 행렬과 길이 2 벡터를 곱하면 길이 3 벡터가 나옵니다.

```python
x = torch.full((2,), 2)
A = torch.full((3, 2), 3)
result = torch.matmul(A, x)
print(result, result.shape)
```

## 4. 배치 matmul — (4, 3, 2) × (2,)

앞쪽 축은 **배치**처럼 여러 장을 한꺼번에 곱합니다. 결과 shape은 (4, 3)입니다.

```python
x = torch.full((2,), 2)
A = torch.full((4, 3, 2), 3)
result = torch.matmul(A, x)
print(result, result.shape)
```

## 5. dtype — NumPy와 텐서

NumPy 배열을 `torch.tensor`로 옮기면 dtype이 함께 따라옵니다.

```python
a = np.ones((2, 3))
b = torch.tensor(a)
print(a.dtype, b.dtype)
```

## 6. reshape · from_numpy · numpy()

모양을 바꾸고, NumPy와 텐서를 **왕복**합니다. NumPy는 기본 실수가 64비트인 경우가 많습니다.

```python
a = torch.tensor([0, 1, 2, 3, 4])
print(a.size(), a.dtype, a.ndim)
print(a.reshape(5, 1))

n_array = np.array([0.0, 1.0, 2.0, 3.0, 4.0])
t_tensor = torch.from_numpy(n_array)
print(t_tensor)
back_numpy = t_tensor.numpy()
print(back_numpy)
```

## 7. item — 원소 하나를 파이썬 숫자로

텐서 원소가 **정확히 1개**일 때만 `.item()`으로 float을 꺼냅니다.

```python
x = torch.randn(1)
print(x)
print(x.item())

v = torch.tensor([1.0, 5.0, 6.0])
print(v[0], v[0].item())
```

## 8. linspace — 구간을 균등하게 나누기

시작·끝·개수를 정하면 NumPy·PyTorch 모두 같은 간격 배열을 만듭니다.

```python
print(np.linspace(-2, 2, 5))
print(torch.linspace(-2, 2, 5))
```

## 9. mm · matmul — 2D 행렬곱

2차원 행렬끼리는 `torch.mm`과 `torch.matmul` 결과가 같습니다.

```python
x = torch.tensor([[0, 1, 1], [1, 0, 1]])
y = torch.tensor([[1, 1], [1, 1], [1, 1]])
print(torch.mm(x, y))
print(torch.matmul(x, y))
```

## 10. sum(dim=...) — 축 따라 합

`dim=0`은 **세로**(열 방향) 합, `dim=1`은 **가로**(행 방향) 합입니다.

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(torch.sum(x, dim=0))
print(torch.sum(x, dim=1))
```

## 11. argmax · argmin — 최댓값·최솟값 위치

값이 아니라 **인덱스**를 돌려줍니다.

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(torch.argmax(x), torch.argmin(x))
```
