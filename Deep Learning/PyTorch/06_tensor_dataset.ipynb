# PyTorch 입문 — TensorDataset · DataLoader

NumPy 배열을 텐서로 바꾼 뒤 `TensorDataset`으로 짝 짓고, `DataLoader`로 배치를 꺼냅니다.  
이론 복습: [`../session06.md`](../session06.md)

> **이 노트북이 하는 일**  
> NumPy 난수 → `from_numpy` → `TensorDataset` → `DataLoader(batch_size=2)` → 한 배치 shape.

환경: `torch-env` 커널을 쓰는 것을 권장합니다.

## 1. import · seed

`TensorDataset`은 여러 텐서를 **같은 인덱스**끼리 묶어 Dataset처럼 씁니다.

```python
from torch.utils.data import TensorDataset, DataLoader
import torch
import numpy as np

# 시드 고정: 다시 실행해도 같은 난수가 나오게 함 (라이브러리별로 따로 고정)
torch.manual_seed(0)  # PyTorch 난수 (torch.randn 등)
np.random.seed(0)  # NumPy 난수 (아래 np.random.randn / randint)
```

## 2. NumPy → 텐서

특징 `X`는 5행 3열, 정답 `Y`는 0/1 다섯 개입니다.  
`from_numpy`로 텐서로 바꿉니다. (모델 입력은 텐서여야 합니다.)

```python
X = np.random.randn(5, 3) # 5행 3열의 랜덤 데이터 생성
Y = np.random.randint(0, 2, (5,)) # 0 이상, 2 미만의 랜덤 5개 데이터 생성

x_torch = torch.from_numpy(X) # X를 텐서로 변환
y_torch = torch.from_numpy(Y) # Y를 텐서로 변환

print("x shape:", x_torch.shape, "y shape:", y_torch.shape) # x_torch의 크기와 y_torch의 크기를 출력
```

## 3. TensorDataset — 인덱스끼리 짝

`train_xy[0]`은 `(x_torch[0], y_torch[0])`입니다.

```python
train_xy = TensorDataset(x_torch, y_torch) # x_torch와 y_torch를 묶어서 train_xy 객체 생성
# train_xy의 값
# [x00, x01, x02], [y0]
# [x10, x11, x12], [y1]
# [x20, x21, x22], [y2]
# [x30, x31, x32], [y3]
# [x40, x41, x42], [y4]
sample_x, sample_y = train_xy[0] # train_xy 객체의 0번째 인덱스를 출력 -> [x00, x01, x02], [y0]
print("샘플 0 x:", sample_x) # sample_x를 출력
print("샘플 0 y:", sample_y) # sample_y를 출력
print("개수:", len(train_xy)) # train_xy 객체의 길이를 출력
```

## 4. DataLoader — batch_size=2 · shuffle

샘플 5개·`batch_size=2`면 대략 2+2+1 묶음입니다.  
`shuffle=True`는 순서를 섞습니다.

```python
dataloader = DataLoader(train_xy, batch_size=2, shuffle=True) # train_xy를 섞고, 2개행씩 묶어서 반환
# 배치 0
# features: [[x00, x01, x02],
#            [x10, x11, x12]]      shape (2, 3)
# labels:   [y0, y1]               shape (2,)

# 배치 1
# features: [[x20, x21, x22],
#            [x30, x31, x32]]      shape (2, 3)
# labels:   [y2, y3]               shape (2,)

# 배치 2  (마지막은 1개만)
# features: [[x40, x41, x42]]      shape (1, 3)
# labels:   [y4]                   shape (1,)
train_features, train_labels = next(iter(dataloader)) # dataloader를 이터레이터로 변환하고, 다음 요소를 반환
# train_features: [[x00, x01, x02], [x10, x11, x12]] shape (2, 3)
# train_labels:   [y0, y1]               shape (2,)
print("batch features shape:", train_features.shape)
print("batch labels shape:", train_labels.shape)
```
