# PyTorch 입문 — TensorDataset · DataLoader

NumPy 배열을 텐서로 바꾼 뒤 `TensorDataset`으로 짝 짓고, `DataLoader`로 배치를 꺼냅니다.  
이론 복습: [`../session06.md`](../session06.md)

> **이 노트북이 하는 일**  
> NumPy 난수 → `from_numpy` → `TensorDataset` → `DataLoader(batch_size=2)` → 한 배치 shape.

환경: `torch-env` 커널을 쓰는 것을 권장합니다.

## 1. import · seed

`TensorDataset`은 여러 텐서를 **같은 인덱스**끼리 묶어 Dataset처럼 씁니다.

```python
from torch.utils.data import TensorDataset, DataLoader
import torch
import numpy as np

torch.manual_seed(0)
np.random.seed(0)
```

## 2. NumPy → 텐서

특징 `X`는 5행 3열, 정답 `Y`는 0/1 다섯 개입니다.  
`from_numpy`로 텐서로 바꿉니다. (모델 입력은 텐서여야 합니다.)

```python
X = np.random.randn(5, 3)
Y = np.random.randint(0, 2, (5,))

x_torch = torch.from_numpy(X)
y_torch = torch.from_numpy(Y)

print("x shape:", x_torch.shape, "y shape:", y_torch.shape)
```

## 3. TensorDataset — 인덱스끼리 짝

`train_xy[0]`은 `(x_torch[0], y_torch[0])`입니다.

```python
train_xy = TensorDataset(x_torch, y_torch)
sample_x, sample_y = train_xy[0]
print("샘플 0 x:", sample_x)
print("샘플 0 y:", sample_y)
print("개수:", len(train_xy))
```

## 4. DataLoader — batch_size=2 · shuffle

샘플 5개·`batch_size=2`면 대략 2+2+1 묶음입니다.  
`shuffle=True`는 순서를 섞습니다.

```python
dataloader = DataLoader(train_xy, batch_size=2, shuffle=True)
train_features, train_labels = next(iter(dataloader))

print("batch features shape:", train_features.shape)
print("batch labels shape:", train_labels.shape)
```
