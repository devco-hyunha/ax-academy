# PyTorch 입문 — nn.Linear 학습 루프

`nn.Linear(1, 1)` · `MSELoss` · `SGD`로 짧은 epoch를 돌리고, 학습 전후 `state_dict`를 비교합니다.  
이론 복습: [`../session07.md`](../session07.md)

> **이 노트북이 하는 일**  
> 가짜 직선 데이터 → `Linear` 모델 → `zero_grad` · `backward` · `step` → 학습 전후 가중치 확인.

환경: `torch-env` 커널을 쓰는 것을 권장합니다.

## 1. import · seed

결과를 같게 맞추려고 seed를 고정합니다.

```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(0)
```

## 2. 가짜 직선 데이터

정답은 `y = 2x + 1` 입니다.  
`unsqueeze(1)`로 열 벡터 모양(샘플 수, 특징 1)을 만듭니다.

```python
x = torch.linspace(-1.0, 1.0, 20).unsqueeze(1)
y = 2.0 * x + 1.0

print("x shape:", x.shape, "y shape:", y.shape)
```

## 3. 모델 · 손실 · Optimizer

`nn.Linear(1, 1)` = 입력 1 → 출력 1인 직선.  
학습 전 `state_dict`는 랜덤 초기값입니다.

```python
model = nn.Linear(1, 1)
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

print("학습 전 state_dict:")
print(model.state_dict())
```

## 4. 학습 루프 — 3연타

매 epoch마다 `zero_grad` → `backward` → `step` 순서를 지킵니다.  
`backward`는 기울기만 구하고, `step`이 가중치를 고칩니다.

```python
for epoch in range(50):
    y_pred = model(x)
    loss = criterion(y_pred, y)

    optimizer.zero_grad()  # 이전 기울기 비우기
    loss.backward()
    optimizer.step()  # weight · bias 갱신

    if epoch % 10 == 0 or epoch == 49:
        print(f"epoch {epoch:2d} | loss={loss.item():.6f}")
```

## 5. 학습 후 state_dict

목표에 가깝면 weight ≈ 2, bias ≈ 1 입니다.

```python
print("학습 후 state_dict:")
print(model.state_dict())
print("목표: weight≈2, bias≈1")
```
