# 유방암 데이터 — 결측 확인

목표는 **악성/양성**을 가리는 분류입니다.  
먼저 데이터가 비어 있는 칸(**결측**)이 있는지 봅니다. 빈칸이 있으면 모델이 헷갈립니다.


```python
# 유방암 표 데이터와 표 도구 불러오기
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np

# load_breast_cancer = 연습용 유방암 숫자 표 가져오기
dataset = load_breast_cancer()

# 표에 쓸 값·정답·칸 이름 꺼내기
# data = 예측에 쓸 숫자들 / target = 정답(악성·양성) / feature_names = 칸 이름
data = dataset['data']
target = dataset['target']
feature_names = dataset['feature_names']

```

```python
# DataFrame = 엑셀처럼 칸 이름 달린 표
# data 숫자를 feature_names 칸 이름으로 붙임
dataframe = pd.DataFrame(data, columns = feature_names)

```

## 결측이란?

**결측** = 값이 비어 있는 칸입니다.  
isnull/isna로 세어 보면 “비어 있는 칸이 몇 개인지”가 나옵니다.


```python
# 일부러 빈칸(결측) 만들기
# iloc = 줄·칸 번호로 표 칸 고르기 (0부터). 앞 5줄의 첫 칸을 빈칸으로
dataframe.iloc[0:5, 0] = np.nan

# fillna = 빈칸 채우기. 여기선 그 칸의 평균값으로 채움
# inplace=True = 새 표를 만들지 않고 지금 표를 바로 고침
dataframe.fillna(dataframe.mean(), inplace = True)
# isna().sum() = 칸마다 빈칸이 몇 개인지 세기 (채운 뒤라 0이어야 함)
dataframe.isna().sum()

```
